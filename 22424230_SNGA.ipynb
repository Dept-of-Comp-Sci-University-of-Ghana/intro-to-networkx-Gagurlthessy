```python
import matplotlib as plt
import networkx as nx
```

```python
DG = nx.DiGraph()
#add the relationships
edges = [("Happy","Theresa"),("Theresa","Happy"),("Happy","Cephas"),("Cephas","Happy"),("Happy","Cephas"),("Cephas","Esther"),("Esther","Bernard")]
DG.add_edges_from(edges)
#customize the networks
nx.draw( DG, node_color = "green", node_size = 1500, arrows = True, with_labels = True, font_color = "white", font_size = 12)
```
